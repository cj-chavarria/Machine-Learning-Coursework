```python
import matplotlib.pyplot as plt
from sklearn.datasets import make_moons, make_circles
from sklearn.metrics import calinski_harabasz_score, adjusted_rand_score

from dbscan_model import DBSCAN
```

```python
X_moons, y_moons = make_moons(n_samples=1000, noise=0.1, random_state=42)
X_circles, y_circles = make_circles(n_samples=1000, noise=0.075, factor=0.5, random_state=42)
```

## Probando el algoritmo

Moons

```python
moons_cluster = DBSCAN(epsilon=0.13, min_pts=5)
moons_pred_labels = moons_cluster.fit(X_moons)

fig, (ax1, ax2) = plt.subplots(1,2, figsize=(15,5))

ax1.scatter(X_moons[:, 0], X_moons[:,1], s=15)
ax1.set_title('Unclustered')

ax2.scatter(X_moons[:, 0], X_moons[:,1], s=15, c=moons_pred_labels, cmap='viridis', edgecolor='k')
ax2.set_title('DBSCAN clustering')

plt.show()

print('Adjust Rand Index score: ', adjusted_rand_score(labels_true=y_moons, labels_pred=moons_pred_labels))
print('Calinski-Harabasz score: ', calinski_harabasz_score(X_moons, moons_pred_labels))
```

Circles

```python
circles_cluster = DBSCAN(epsilon=0.13, min_pts=5)
circles_pred_labels = circles_cluster.fit(X_circles)

fig, (ax1, ax2) = plt.subplots(1,2, figsize=(15,5))

ax1.scatter(X_circles[:, 0], X_circles[:,1], s=10)
ax1.set_title('Unclustered')

ax2.scatter(X_circles[:,0], X_circles[:,1], s=10, c=circles_pred_labels, cmap='plasma', edgecolor='k')
ax2.set_title('DBSCAN clustering')

plt.show()

print('Adjust Rand Index score: ', adjusted_rand_score(labels_true=y_circles, labels_pred=circles_pred_labels))
print('Calinski-Harabasz score: ', calinski_harabasz_score(X_circles, circles_pred_labels))
```

### Comparacion con KMeans

```python
from sklearn.cluster import KMeans
```

Moons

```python
moons_kmeans_pred_labels = KMeans(n_clusters=2, max_iter=1000).fit_predict(X_moons)

fig, (ax1, ax2) = plt.subplots(1,2, figsize=(15,5))

ax1.scatter(X_moons[:,0], X_moons[:,1], s=10, c=moons_kmeans_pred_labels, cmap='viridis', edgecolor='k')
ax1.set_title('KMeans')
ax2.scatter(X_moons[:, 0], X_moons[:,1], s=15, c=moons_pred_labels, cmap='viridis', edgecolor='k')
ax2.set_title('DBSCAN')

plt.show()

print('KMeans Adjust Rand Index score: ', adjusted_rand_score(labels_true=y_moons, labels_pred=moons_kmeans_pred_labels))
print('DBSCAN Adjust Rand Index score: ', adjusted_rand_score(labels_true=y_moons, labels_pred=moons_pred_labels))
print('')
print('KMeans Calinski-Harabasz score: ', calinski_harabasz_score(X_moons, moons_kmeans_pred_labels))
print('DBSCAN Calinski-Harabasz score: ', calinski_harabasz_score(X_moons, moons_pred_labels))
```

Circles

```python
circles_kmeans_pred_labels = KMeans(n_clusters=2, max_iter=1000).fit_predict(X_circles)

fig, (ax1, ax2) = plt.subplots(1,2, figsize=(15,5))

ax1.scatter(X_circles[:,0], X_circles[:,1], s=10, c=circles_kmeans_pred_labels, cmap='plasma', edgecolor='k')
ax1.set_title('KMeans')
ax2.scatter(X_circles[:, 0], X_circles[:,1], s=15, c=circles_pred_labels, cmap='plasma', edgecolor='k')
ax2.set_title('DBSCAN')

plt.show()

print('KMeans Adjust Rand Index score: ', adjusted_rand_score(labels_true=y_circles, labels_pred=circles_kmeans_pred_labels))
print('DBSCAN Adjust Rand Index score: ', adjusted_rand_score(labels_true=y_circles, labels_pred=circles_pred_labels))
print('')
print('KMeans Calinski-Harabasz score: ', calinski_harabasz_score(X_circles, circles_kmeans_pred_labels))
print('DBSCAN Calinski-Harabasz score: ', calinski_harabasz_score(X_circles, circles_pred_labels))
```

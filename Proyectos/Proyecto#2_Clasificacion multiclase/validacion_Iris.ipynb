# Validacion usando Iris dataset

```python
from sklearn.model_selection import cross_val_score
from sklearn.datasets import load_iris
from modelos import OneVsAll, OneVsOne, SoftMaxClassifier
```

```python
iris = load_iris()
```

```python
X = iris.data
y = iris.target
```

### One vs All

```python
OvA_log = OneVsAll(bin_clf='logistic_reg')
OvA_sgd = OneVsAll(bin_clf='sgd')
```

```python
OvA_log_score = cross_val_score(OvA_log, X, y, cv=5, n_jobs=-1, scoring='accuracy')
OvA_sgd_score = cross_val_score(OvA_sgd, X, y, cv=5, n_jobs=-1, scoring='accuracy')
```

```python
print('OneVsAll (Logistic) score: ', OvA_log_score.mean())
print('OneVsAll (SGD) score: ', OvA_sgd_score.mean())
```

### One Vs One

```python
OvO_log = OneVsOne(bin_clf='logistic_reg')
OvO_sgd = OneVsOne(bin_clf='sgd')
```

```python
OvO_log_score = cross_val_score(OvO_log, X, y, cv=5, n_jobs=-1, scoring='accuracy')
OvO_sgd_score = cross_val_score(OvO_sgd, X, y, cv=5, n_jobs=-1, scoring='accuracy')
```

```python
print('OneVsOne (Logistic) score: ', OvO_log_score.mean())
print('OneVsOne (SGD) score: ', OvO_sgd_score.mean())
```

### SoftMax Regression

```python
softmax = SoftMaxClassifier()
```

```python
softmax_score = cross_val_score(softmax, X, y, cv=5, n_jobs=-1, scoring='accuracy')
```

```python
print('SoftMax score: ', softmax_score.mean())
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
plt.rc('font', size=14)
plt.rc('axes', labelsize=14, titlesize=14)
plt.rc('legend', fontsize=14)
plt.rc('xtick', labelsize=10)
plt.rc('ytick', labelsize=10)
```

```python
def sigmoid(t):
    return 1 / (1 + np.exp(-t))
```

```python
t = np.linspace(-10, 10, 50)
sig_t = sigmoid(t)
```

```python
plt.plot(t, sig_t, 'b-', linewidth=2, label=r'$\sigma(t) = \dfrac{1}{1+e^{-t}}$')
plt.xlabel('t')
plt.ylabel(r'$\sigma(t)$')
plt.legend(loc='upper left', framealpha=1.0)
plt.gca().set_yticks([0,0.25, 0.5,0.75,1])
plt.grid()
plt.show()
```

```python
X = np.array([0.052, 0.081, 0.096, 0.158, 0.242, 0.253, 0.396, 0.427, 0.459, 0.572])
y = np.array([0, 0, 0, 0, 1, 0, 1, 1, 1, 1])
```

```python
m = X.size
theta1 =  np.arange(-40,60, 0.1)
h =  np.array([-0.5 + theta*X for theta in theta1])
J = (1/m)*((h-y)**2).sum(axis=1) #MSE

plt.plot(theta1, J, 'b-', linewidth=2)

plt.xlabel(r'$\theta$', fontsize=12)
plt.ylabel(r'$J(\theta)$', fontsize=12)
plt.xlim(theta1.min(), theta1.max())
plt.grid()
plt.show()
```

```python
m = X.size
theta1 =  np.arange(-40,60, 0.1)
h =  np.array([-0.5 + theta*X for theta in theta1])
h = sigmoid(h)
J = (1/m)*((h-y)**2).sum(axis=1) #MSE

plt.plot(theta1, J, 'b-', linewidth=2)

plt.xlabel(r'$\theta$', fontsize=12)
plt.ylabel(r'$J(\theta)$', fontsize=12)
plt.xlim(theta1.min(), theta1.max())
plt.grid()
plt.show()
```

```python
p = np.arange(0.01, 1, 0.01)
c = -np.log(p)

plt.plot(p, c, 'b-')
plt.xlabel(r'$\hat{p}$')
plt.ylabel(r'$c(\theta)$')
plt.title(r'Cost function when class label $y=1$')
plt.xlim(0,1)
plt.grid()
plt.show()
```

```python
p = np.arange(0.01, 1, 0.01)
c = -np.log(1-p)

plt.plot(p, c, 'b-')
plt.xlabel(r'$\hat{p}$')
plt.ylabel(r'$c(\theta)$')
plt.title(r'Cost function when class label $y=0$')
plt.xlim(0,1)
plt.grid()
plt.show()
```

```python
m = X.size
theta1 =  np.arange(-40,60, 0.1)
h =  np.array([-0.5 + theta*X for theta in theta1])
h = sigmoid(h)
J = -(1/m)*(y*np.log(h) + (1-y)*np.log(1-h)).sum(axis=1) #Logistic Regression cost function

plt.plot(theta1, J, 'b-', linewidth=2)

plt.xlabel(r'$\theta$', fontsize=12)
plt.ylabel(r'$J(\theta)$', fontsize=12)
plt.xlim(theta1.min(), theta1.max())
plt.grid()
plt.show()
```

```python
from sklearn.preprocessing import add_dummy_feature
X = iris.data[:, 3:]
X_b =  add_dummy_feature(X)
```

```python
y = (iris.target==2).astype(int)[:, np.newaxis]
y
```

```python
m = X.shape[0]
eta = 10
n_epochs = 10000
```

```python
rng = np.random.default_rng(seed=42)
```

```python
theta =  rng.standard_normal((2,1))
theta
```

```python
from numpy import gradient


for i in range(n_epochs):
    gradients = 1/m * X_b.T@(sigmoid(X_b@theta) - y)
    theta -= gradients

theta
```

```python
X_new = np.linspace(0,3,1000)[:, np.newaxis]
X_new_b = add_dummy_feature(X_new)
y_proba = sigmoid(X_new_b@theta)

decision_bundary =  X_new[y_proba >= 0.5][0]

print(f'decision_bundary; {decision_bundary}')

plt.figure(figsize=(8,5))
plt.plot(X[y==0], y[y==0], 'bs')
plt.plot(X[y==1], y[y==1], 'g^')
plt.plot([decision_bundary, decision_bundary], [-1,2], 'k:', linewidth=2, label='Decision Bundary')

plt.plot(X_new, y_proba, 'g-', label='Iris Virginica')
plt.plot(X_new, 1-y_proba, 'b--', label='Not Iris Virginica')

plt.xlabel('Petal width (cm)')
plt.ylabel('Probability')
plt.legend(loc='upper left')
plt.axis([0,3,-0.02, 1.02])
plt.grid()
plt.show()
```

```python
X_new[y_proba >= 0.5]
```

```python
from sklearn.linear_model import LogisticRegression

model = LogisticRegression()
model.fit(X, y.ravel())
```

```python
model.predict_proba([[0.8], [1.5], [1.8]])
```

```python
model.predict([[0.8], [1.5], [1.8]])
```

```python
X_new = np.linspace(0,3,1000)[:, np.newaxis]
y_proba = model.predict_proba(X_new)[:, 1:]

decision_bundary =  X_new[y_proba >= 0.5][0]

print(f'decision_bundary; {decision_bundary}')

plt.figure(figsize=(8,5))
plt.plot(X[y==0], y[y==0], 'bs')
plt.plot(X[y==1], y[y==1], 'g^')
plt.plot([decision_bundary, decision_bundary], [-1,2], 'k:', linewidth=2, label='Decision Bundary')

plt.plot(X_new, y_proba, 'g-', label='Iris Virginica')
plt.plot(X_new, 1-y_proba, 'b--', label='Not Iris Virginica')

plt.xlabel('Petal width (cm)')
plt.ylabel('Probability')
plt.legend(loc='upper left')
plt.axis([0,3,-0.02, 1.02])
plt.grid()
plt.show()
```

```python
X =  iris.data[:, (2,3)]
y = (iris.target == 2).astype(int)
```

```python
model =  LogisticRegression(C=10**10)
model.fit(X, y)
```

```python
left_right = np.array([0, 7])

boundary = -(model.intercept_[0] + model.coef_[0][0]*left_right)/model.coef_[0][1]

plt.figure(figsize=(8,5))
plt.plot(left_right, boundary, 'k:', linewidth=2, label='Decision Boundary')
plt.plot(X[y==0,0], X[y==0,1], 'bs', label='Not Iris Virginica')
plt.plot(X[y==1,0], X[y==1,1], 'g^', label='Iris Virginica')

plt.xlabel('Petal length (cm)')
plt.ylabel('Petal width (cm)')
plt.legend(loc='center left')
plt.axis([0,7,0,3.5])
plt.grid()
plt.show()
```

```python
X = iris.data[:, (2,3)]
y = (iris.target == 2).astype(int)

model = LogisticRegression(C=10*10)
model.fit(X, y)

left_right = np.array([0, 7])
boundary = -(model.intercept_[0] + model.coef_[0][0]*left_right) / model.coef_[0][1]

x0, x1 = np.meshgrid(
  np.linspace(0,7,500)[:, np.newaxis],
  np.linspace(0,3.5,500)[:, np.newaxis],
)

X_new = np.c_[x0.ravel(), x1.ravel()]

y_predict = model.predict(X_new)

zz = y_predict.reshape(x0.shape)

plt.figure(figsize=(8,5))
plt.plot(left_right, boundary, "k:", linewidth=2, label="Decision boundary")

plt.plot(X[y==0, 0], X[y==0, 1], "bs", label="Iris Virginica")
plt.plot(X[y==1, 0], X[y==1, 1], "g^", label="No Iris Virginica")

from matplotlib.colors import ListedColormap

custom_map = ListedColormap(["#fafab0", "#9898ff", "#a0faa0"])
plt.contourf(x0, x1, zz, cmap=custom_map)

plt.xlabel("Petal legnth (cm)")
plt.xlabel("Petal width (cm)")

plt.legend(loc="center left")
plt.axis([0, 7, 0, 3.5])
plt.show()
```
